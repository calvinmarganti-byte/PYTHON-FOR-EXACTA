## **Matrix Operation and Transpose**

*   Addition-Substraction
*   Multiplication-Division
*   Scalars





```python
#MATRIX MULTIPLICATION
import numpy as np


A = np.array ([
    [1, 2, 3], [5, 3, 2]
])

B = np.array ([
    [5, 2, 10], [3, 2, 7], [3,3,3]
])

print(A)
print (B)
print("perkalian dari", A, "x", B, "adalah:", A@B )

#If A = m*n, B must be n*k with m, n, and k are integers

```

```python
#Matrix Division
import numpy as np

A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print("Pembagian dari matrix A dan B adalah", A @ np.linalg.inv(B))
```

```python
#MATRIX ADDITION
A = np.array ([
    [1, 2, 3], [5, 3, 2], [5,2,1]
])

B = np.array ([
    [5, 2, 10], [3, 2, 7], [3,3,3]
])

print(A)
print (B)
print("penjumlahan dari", A, "x", B, "adalah:", A+B )
```

```python
#SCALAR AND MATRIX
A = np.array ([
    [1, 2, 3], [5, 3, 2], [5,2,1]
])

print(A*10)
```

# Determinan and Invers

```python
import numpy as np

A = ([
    [1, 2, 3], [2,3,4], [5,6,7]
])
det = np.linalg.det(A)
print("Maka determinan ialah", det)
inv = np.linalg.inv(A)
print("Maka invers dari matriks ini ialah", inv)
```

# Eigenvalues and Eigenvector

```python
import numpy as np

A = ([
    [1, 2, 3], [2,3,4], [5,6,7]
])

values, vectors = np.linalg.eig(A)

print("Ini adalah eigenvalues:", values)
print("Ini adalah eigenvectors", vectors)
```

# Rank Matrix

```python
import numpy as np

A = ([
    [1, 2, 3], [2, 4, 7]
])

rank = np.linalg.matrix_rank(A)
print("Maka, matrix rank dari matrix ini adalah", rank)
```

# Linear Algebra Problem Solving for 2 Variables



> Given such linear problem:
x + 2y = 5
3x + 4y = 11




```python
import numpy as np

A = np.array ([
    [1, 2], [3, 4]
])

B = np.array ([5, 11])

solution = np.linalg.solve(A, B)

print(solution)
```
